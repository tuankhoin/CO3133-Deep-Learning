# 1. Basic Concepts
---

## What is Deep Learning?

<img src="https://github.com/tuankhoin/CO3057-Computer-Vision/blob/main/assets/m_sequence.JPG?raw=true" height=300/>

**Deep learning** is a subset of machine learning that uses artificial neural networks with multiple layers to learn representations and make decisions from data.

Key characteristics:
- Uses neural networks with many layers — hence *deep*
- Automatically learns hierarchical representations from data
- Can work directly with raw inputs such as images, text, and audio
- Has achieved strong performance across many application domains

A useful high-level view is:

$$
\text{input data} \rightarrow \text{learned representations} \rightarrow \text{prediction}
$$

## Traditional Machine Learning vs Deep Learning

![](https://static.vietnix.vn/wp-content/uploads/2022/04/deep-learning-hoat-dong-nhu-the-nao.webp)

| Aspect | Traditional Machine Learning | Deep Learning |
|---|---|---|
| Feature engineering | Often manually designed | Features learned automatically |
| Data requirements | Can work well with smaller datasets | Often benefits strongly from large datasets |
| Computation | Usually less intensive | Commonly uses GPUs/TPUs |
| Interpretability | Often easier to interpret | Often less interpretable |
| Typical strengths | Structured/tabular data | Images, text, audio and other unstructured data |
| Examples | Linear regression, SVM, Random Forest | CNNs, RNNs, Transformers |

**Key idea:** traditional ML often separates *feature extraction* from *prediction*, whereas deep learning often learns both together.

## Historical Milestones

| Period | Milestones |
|---|---|
| **Early foundations** | 1943: McCulloch-Pitts neuron; 1958: Rosenblatt's Perceptron; 1986: backpropagation popularized by Rumelhart, Hinton & Williams; 1989: universal approximation results |
| **Deep learning revolution** | 2006: Deep Belief Networks; 2012: AlexNet wins ImageNet; 2014: GANs; 2017: Transformer |
| **Large-scale foundation models** | 2018-2022: BERT, GPT and large language models |
| **Multimodal AI** | 2023-present: systems combining text, images, audio and/or video |

The broad trend is increasing **model capacity + data + computation + scalable training methods**.

 x' = \frac{x-\mu}{\sigma}

## Data Representation: Text and Audio

### Text
- **Character-level:** each character is a token
- **Word-level:** each word is a token
- **Subword-level:** BPE / WordPiece, common in BERT- and GPT-style systems
- **One-hot encoding:** sparse binary vectors
- **Word embeddings:** dense vectors such as Word2Vec / GloVe
- **Contextual embeddings:** representation depends on context, as in BERT

Typical quantities include vocabulary size $V$, embedding dimension $d$, and sequence length $L$.

### Audio
- **Raw waveform:** time-domain signal, shape $(T,)$
- **Spectrogram:** time-frequency representation, shape $(F,T)$
- **Mel-spectrogram:** perceptually motivated frequency scale
- **MFCC:** Mel-frequency cepstral coefficients

Typical preprocessing includes selecting a sampling rate, window/frame size, hop size, and normalization.

## Main Problem Types

### Unsupervised / Self-Supervised Learning

- **Clustering:** group similar samples
- **Dimensionality reduction:** learn compact representations
- **Generative modeling:** learn the data distribution $p(x)$ and generate samples
  - VAE, GAN, diffusion models
- **Self-supervised learning:** construct learning signals from unlabeled data

### Reinforcement Learning

<img src="https://vcdn1-ngoisao.vnecdn.net/2010/10/05/dv3-254923-1368147048.jpg?w=300&h=180&q=100&dpr=2&fit=crop&s=1Cl2oZUodhWGV7tWZZkAmQ" height=150/>

- Learn a policy through interaction and reward
- Examples: DQN, policy gradients, actor-critic

## Evaluation Metrics: Classification

For binary classification:

| Metric | Formula | Useful When |
|---|---|---|
| Accuracy | $\frac{TP+TN}{TP+TN+FP+FN}$ | classes are reasonably balanced |
| Precision | $\frac{TP}{TP+FP}$ | false positives are costly |
| Recall | $\frac{TP}{TP+FN}$ | false negatives are costly |
| F1 | $2\frac{\text{Precision}\cdot\text{Recall}}{\text{Precision}+\text{Recall}}$ | balance precision and recall |
| ROC-AUC | area under ROC curve | ranking quality across thresholds |

For multi-class classification, common reports include:
- Per-class Precision / Recall / F1
- **Macro** average: each class has equal weight
- **Weighted** average: classes weighted by frequency
- **Micro** average: aggregate $TP$, $FP$, and $FN$ first

## Evaluation Metrics: Regression, Detection, and Segmentation

### Regression
| Metric | Formula / Meaning |
|---|---|
| MAE | $\frac{1}{n}\sum_i |y_i-\hat y_i|$ |
| MSE | $\frac{1}{n}\sum_i (y_i-\hat y_i)^2$ |
| RMSE | $\sqrt{\frac{1}{n}\sum_i (y_i-\hat y_i)^2}$ |
| $R^2$ | $1-\frac{\sum_i(y_i-\hat y_i)^2}{\sum_i(y_i-\bar y)^2}$ |

### Object Detection
<img src="https://www.baeldung.com/wp-content/uploads/sites/4/2022/04/fig1.png" height=200/>

- **IoU:** $\frac{\text{Area of Intersection}}{\text{Area of Union}}$
- **AP / mAP:** precision-recall performance, often evaluated at one or multiple IoU thresholds

### Segmentation
- Pixel accuracy
- Mean IoU (mIoU)
- Dice coefficient

$$
\text{Dice}(X,Y)=\frac{2|X\cap Y|}{|X|+|Y|}
$$

# 4. Data Analysis Workflow
---

## Step 1-3: Define, Collect, Prepare

### 1. Problem Definition
Ask:
- What is the task: classification, regression, detection, generation, ...?
- What constraints matter: accuracy, latency, memory, compute?
- What metric defines success?
- What data is available: size, quality, format?

### 2. Data Collection & Preparation
Possible sources:
- Public datasets: ImageNet, COCO, CIFAR, MNIST
- Private/company data
- Custom data collection
- Synthetic / augmented data
- Web collection where permitted

Preparation includes cleaning, annotation, format conversion, and quality checking.

### 3. Data Preprocessing
- Images: resize, normalize, augment, tensor conversion
- Text: tokenize, encode, pad/truncate, embed

## Step 4-5: Design and Train

### 4. Model Design & Selection
- Choose a task-appropriate architecture
  - CNNs for many image tasks
  - RNNs / Transformers for sequences
- Consider pretrained models and transfer learning
- Balance model size against speed and memory

Common hyperparameters:
- Learning rate
- Batch size
- Number of layers / hidden units
- Dropout rate / weight decay

### 5. Training
1. Initialize weights — random or pretrained
2. Forward pass
3. Compute loss
4. Backward pass via backpropagation
5. Optimizer updates parameters
6. Repeat for multiple epochs

Best practices:
- Monitor training and validation curves
- Use checkpoints
- Consider early stopping and learning-rate scheduling

## Step 6-7: Evaluate, Deploy, Monitor

### 6. Evaluation & Validation
- Use the **validation set** for tuning and model selection
- Use the **test set** for final evaluation
- Compare architectures and hyperparameters consistently
- Perform error analysis to understand failure cases

### 7. Deployment & Monitoring
Deployment options:
- Model optimization: quantization, pruning, distillation
- Model serving: APIs / serving systems
- Edge or embedded deployment
- Cloud deployment

Monitor after deployment:
- Accuracy / task metrics
- Latency
- Data drift
- Model degradation
- Need for retraining or A/B testing

# 13. Performance Metrics & Model Evaluation
---

<img src="https://miro.medium.com/v2/resize:fit:1400/1*GhKMAUmi4bfFiEwZCPlDsA.png" height=300/>

## Evaluation Techniques

| Technique | Main Idea | Typical Use |
|---|---|---|
| **Hold-out** | split into training / validation / test | large datasets |
| **$k$-fold cross-validation** | rotate validation fold across $k$ splits | smaller datasets / robust model selection |
| **Bootstrap** | repeatedly resample with replacement | uncertainty estimation |

Important considerations:
- Keep splits representative
- Avoid leakage
- Cross-validation and bootstrap increase computational cost
- Keep a final unseen test set for unbiased evaluation

## $k$-Fold Cross-Validation

1. Split the dataset into $k$ similarly sized folds
2. Train on $k-1$ folds and validate on the remaining fold
3. Repeat $k$ times so each fold is used for validation once
4. Average metrics across folds

Benefits:
- Every sample can contribute to both training and validation across runs
- Gives a more robust estimate than a single split

After model/hyperparameter selection, retrain the selected configuration using the available development data before deployment, while preserving any final held-out test set required for unbiased reporting.

## Bootstrap Evaluation

For a dataset of size $n$:
- Draw a new sample of size $n$ **with replacement**
- Fit/evaluate the model
- Repeat for $B$ bootstrap runs
- Aggregate the mean, standard deviation, or confidence interval of the metric

### Out-of-Bag (OOB) Samples
A sample is omitted from one bootstrap draw with probability approximately

$$
\left(1-\frac{1}{n}\right)^n \approx e^{-1} \approx 0.368
$$

So about **36.8%** of samples are OOB on average and can be used as a natural validation set for that bootstrap model.

Bootstrap is especially useful for small datasets and estimating uncertainty.

# 15. The Bias-Variance Trade-off
---

![](https://media.geeksforgeeks.org/wp-content/uploads/20251212100155481380/70830267-.webp)

## Bias and Variance

- **Bias:** systematic error from a model that is too limited to capture the underlying relationship  
  $\rightarrow$ associated with **underfitting**

- **Variance:** sensitivity to the particular training samples  
  $\rightarrow$ associated with **overfitting**

Practical responses:
- High bias: increase model capacity / improve representation
- High variance: regularize, collect more data, or reduce unnecessary complexity
- Use validation / cross-validation to identify a useful balance

## Bias-Variance Decomposition

For squared prediction error:

$$
\mathbb{E}\left[(y-\hat f(x))^2\right]
=
\underbrace{\left(\mathbb{E}[\hat f(x)]-f(x)\right)^2}_{\text{Bias}^2}
+
\underbrace{\mathbb{E}\left[(\hat f(x)-\mathbb{E}[\hat f(x)])^2\right]}_{\text{Variance}}
+
\underbrace{\sigma^2}_{\text{Noise}}
$$

- **Bias$^2$:** inability of the learned model family/procedure to represent the true relationship
- **Variance:** sensitivity of the learned predictor to changes in the training sample
- **Noise:** irreducible randomness or measurement noise

## Generalization

The goal of machine learning is not to memorize the training set, but to perform well on **unseen data**.

Important observations from the lecture:
- More high-quality and diverse data usually improves generalization
- Data cleaning, normalization, and augmentation can be as important as model choice
- Imbalanced or biased data can lead to poor real-world performance

**Key message:** generalization depends heavily on the quality, diversity, and representativeness of the data.

# 17. Performance Metrics: Detail
---

## Binary Classification: Confusion Matrix

| | Predicted Positive | Predicted Negative |
|---|---:|---:|
| **Actual Positive** | $TP$ | $FN$ |
| **Actual Negative** | $FP$ | $TN$ |

$$
\text{Accuracy}=\frac{TP+TN}{TP+TN+FP+FN}
$$

$$
\text{Precision}=\frac{TP}{TP+FP},\qquad
\text{Recall}=\frac{TP}{TP+FN}
$$

$$
F1=2\frac{\text{Precision}\cdot\text{Recall}}{\text{Precision}+\text{Recall}}
$$

## Multi-Class Classification Metrics

For a $K\times K$ confusion matrix $C=[c_{ij}]$:
- $c_{ij}$: number of true class $i$ samples predicted as class $j$
- Row sum: number of ground-truth samples in class $i$
- Column sum: number of predictions in class $j$
- Diagonal: correct predictions

For class $k$:

$$
\text{Precision}_k=\frac{c_{kk}}{\sum_i c_{ik}},\qquad
\text{Recall}_k=\frac{c_{kk}}{\sum_j c_{kj}}
$$

$$
F1_k=2\frac{\text{Precision}_k\text{Recall}_k}{\text{Precision}_k+\text{Recall}_k}
$$

## Macro, Weighted, and Micro Averaging

### Macro
Each class contributes equally:

$$
\text{Macro-F1}=\frac{1}{K}\sum_{k=1}^{K}F1_k
$$

### Weighted
Weight each class by $w_k=n_k/N$:

$$
\text{Weighted-F1}=\sum_{k=1}^{K}w_kF1_k
$$

### Micro
Aggregate counts first:

$$
\text{Micro-Precision}=
\frac{\sum_k TP_k}{\sum_k(TP_k+FP_k)}
$$

$$
\text{Micro-Recall}=
\frac{\sum_k TP_k}{\sum_k(TP_k+FN_k)}
$$

For standard single-label multi-class classification, micro-averaged precision/recall/F1 coincide with overall accuracy; this is not generally true for multi-label classification.

## ROC and AUC

The ROC curve plots **True Positive Rate (TPR)** against **False Positive Rate (FPR)** while varying the decision threshold:

$$
TPR=\frac{TP}{TP+FN},\qquad
FPR=\frac{FP}{FP+TN}
$$

AUC is the area under this curve:

$$
AUC=\int_0^1 TPR(FPR)\,d(FPR)
$$

A useful interpretation is the probability that a randomly chosen positive sample receives a higher score than a randomly chosen negative sample:

$$
AUC=P(s^+>s^-)
$$

For multi-class problems, ROC-AUC can be extended through one-vs-rest aggregation.

## Regression Metrics: Detail

| Metric | Formula -------------- | Interpretation |
|---|---|---|
| **MAE** | $\frac{1}{n}\sum_i |y_i-\hat y_i|$ | average error magnitude; less sensitive to large outliers than MSE |
| **MSE** | $\frac{1}{n}\sum_i (y_i-\hat y_i)^2$ | penalizes large errors strongly |
| **RMSE** | $\sqrt{\frac{1}{n}\sum_i (y_i-\hat y_i)^2}$ | same units as target |
| **$R^2$** | $1-\frac{\sum_i(y_i-\hat y_i)^2}{\sum_i(y_i-\bar y)^2}$ | fraction of target variance explained relative to mean baseline |

For $R^2$:
- $R^2=1$: perfect prediction
- $R^2=0$: no better than predicting the mean
- $R^2<0$: worse than the constant-mean baseline

# 18. Bias-Variance: Derivation
---

Start with the expected squared prediction error:

$$
\mathbb{E}\left[(y-\hat f(x))^2\right]
$$

Assume

$$
y=f(x)+\epsilon,\qquad \mathbb{E}[\epsilon]=0
$$

where $f(x)$ is the true function and $\epsilon$ is random noise. Substituting gives

$$
\mathbb{E}\left[(f(x)+\epsilon-\hat f(x))^2\right]
$$

## Step 1: Separate the Noise

Expanding:

$$
\mathbb{E}\left[(f(x)-\hat f(x))^2\right]
+2\mathbb{E}\left[(f(x)-\hat f(x))\epsilon\right]
+\mathbb{E}[\epsilon^2]
$$

Assume $\epsilon$ has zero mean and is independent of the predictor. Then the cross-term vanishes:

$$
\mathbb{E}\left[(f(x)-\hat f(x))\epsilon\right]=0
$$

Therefore:

$$
\mathbb{E}\left[(y-\hat f(x))^2\right]
=
\mathbb{E}\left[(f(x)-\hat f(x))^2\right]
+\mathbb{E}[\epsilon^2]
$$

## Step 2: Separate Bias and Variance

Define the expected model prediction

$$
\mu(x)=\mathbb{E}[\hat f(x)]
$$

Add and subtract $\mu(x)$:

$$
\mathbb{E}\left[(f(x)-\mu(x)+\mu(x)-\hat f(x))^2\right]
$$

This separates into:

$$
\underbrace{(f(x)-\mathbb{E}[\hat f(x)])^2}_{\text{Bias}^2}
+
\underbrace{\mathbb{E}\left[(\hat f(x)-\mathbb{E}[\hat f(x)])^2\right]}_{\text{Variance}}
$$

Combining with noise:

$$
\boxed{
\mathbb{E}\left[(y-\hat f(x))^2\right]
=
\text{Bias}^2+\text{Variance}+\sigma^2
}
$$

## Main Ideas to Remember
- Deep learning learns hierarchical representations from data
- The data type and label structure determine how a problem is formulated
- Choose metrics that match the real task objective

- Generalization matters more than training-set performance
